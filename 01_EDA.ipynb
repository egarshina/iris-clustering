```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Загрузка данных
iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names
target_names = iris.target_names

# Создание DataFrame
df = pd.DataFrame(X, columns=feature_names)
df['species'] = [target_names[i] for i in y]

print(df.head())
print(f"\nРазмерность: {df.shape}")
print(f"Классы: {target_names}")
```

```python
df.hist(figsize=(10, 8), bins=20)
plt.suptitle('Распределение признаков', y=0.95)
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.ravel()
for i, col in enumerate(feature_names):
    sns.boxplot(x='species', y=col, data=df, ax=axes[i])
    axes[i].set_title(f'{col} по видам')
plt.tight_layout()
plt.show()
```

```python
sns.pairplot(df, hue='species', diag_kind='hist')
plt.suptitle('Матрица диаграмм рассеяния (Pairplot)', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
# Только числовые признаки
corr = df.iloc[:, :-1].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Матрица корреляций признаков')
plt.show()
```
